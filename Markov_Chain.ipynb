```python
import random
from collections import defaultdict

def build_markov_chain(text):
  words = text.split()
  chain = defaultdict(list)

  for i in range(len(words) - 1):
    chain[words[i]].append(words[i+1])

  return chain


def generate_text(chain, length=30):
  if not chain:
    return ""

  current_word = random.choice(list(chain.keys()))
  output = [current_word]

  for _ in range(length - 1):
    next_words = chain.get(current_word)
    if not next_words:
      current_word = random.choice(list(chain.keys()))
      output.append(current_word)
      continue

    current_word = random.choice(next_words)
    output.append(current_word)

  return " ". join(output)

from gutenberg.acquire import load_etext
from gutenberg.cleanup import strip_headers

# Amusements in Mathematics
text1 = strip_headers(load_etext(16713, mirror='http://www.gutenberg.org/')).strip()
# Anne of Green Gables
text2 = strip_headers(load_etext(45, mirror='http://www.gutenberg.org/')).strip()
# Don Quixote
text3 = strip_headers(load_etext(996, mirror='http://www.gutenberg.org/')).strip()

texts = [text1, text3]

combined_text = " ".join(texts)
combined_text = combined_text.lower()

chain = build_markov_chain(combined_text)

print(generate_text(chain, length = 10))
```
